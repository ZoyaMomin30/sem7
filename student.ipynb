```python
# checking how many passed the exam 
def result(num):
    if num>=33:
        return 'Pass'
    else:
        return 'Fail'
```

```python
df['Result'] = df['exam_score'].apply(result)
```

```python
df['Result'].value_counts()
```

```python
#analysis on students passed or failed.
sns.countplot(df['Result'])
death_percent=round((df['Result'].value_counts().values[0]/891)*100)
print("out of 200 people {}% passed ".format(death_percent))
```
